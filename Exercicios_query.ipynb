2 - Encontre o nome e o endereço dos clientes que fizeram um pedido em 2024.

```python
conn = psycopg2.connect(
            dbname="ecommerce",
            user="postgres",
            password="********",
            host="localhost",
            port="5432"
        )

cursor = conn.cursor()

# Executar a consulta
query = """
SELECT DISTINCT C.Nome, C.Endereco
FROM Clientes C
JOIN Pedidos P ON C.ID_Cliente = P.ID_Cliente
WHERE EXTRACT(YEAR FROM P.Data_Pedido) = 2024;

"""
cursor.execute(query)

# Recuperar e imprimir os resultados
resultados = cursor.fetchall()
for row in resultados:
    print(row)

# Fechar a conexão
cursor.close()
conn.close()
```

```python
conn = psycopg2.connect(
            dbname="ecommerce",
            user="postgres",
            password="********",
            host="localhost",
            port="5432"
        )

cursor = conn.cursor()

# Executar a consulta
query = """
SELECT COUNT(*) AS Total_Pedidos
FROM Pedidos
WHERE EXTRACT(YEAR FROM Data_Pedido) = 2024 AND EXTRACT(MONTH FROM Data_Pedido) = 7;
"""
cursor.execute(query)

# Recuperar e imprimir os resultados
resultados = cursor.fetchall()
for row in resultados:
    print(row)

# Fechar a conexão
cursor.close()
conn.close()
```

8 - Liste os pedidos com seu status e a quantidade de itens em cada pedido.

```python
conn = psycopg2.connect(
            dbname="ecommerce",
            user="postgres",
            password="********",
            host="localhost",
            port="5432"
        )

cursor = conn.cursor()

# Executar a consulta
query = """
SELECT P.ID_Pedido, P.Status, COUNT(IP.ID_Item) AS Quantidade_Itens
FROM Pedidos P
JOIN Itens_Pedido IP ON P.ID_Pedido = IP.ID_Pedido
GROUP BY P.ID_Pedido, P.Status;
"""
cursor.execute(query)

# Recuperar e imprimir os resultados
resultados = cursor.fetchall()
for row in resultados:
    print(row)

# Fechar a conexão
cursor.close()
conn.close()
```

9 - Encontre os produtos que não foram vendidos em nenhum pedido.

```python
conn = psycopg2.connect(
            dbname="ecommerce",
            user="postgres",
            password="********",
            host="localhost",
            port="5432"
        )

cursor = conn.cursor()

# Executar a consulta
query = """
SELECT P.*
FROM Produtos P
LEFT JOIN Itens_Pedido IP ON P.ID_Produto = IP.ID_Produto
WHERE IP.ID_Produto IS NULL;
"""
cursor.execute(query)

# Recuperar e imprimir os resultados
resultados = cursor.fetchall()
for row in resultados:
    print(row)

# Fechar a conexão
cursor.close()
conn.close()
```

10 - Liste o nome dos 10 maiores clientes e a quantidade de pedidos feitos por cada um. (por quantidade de pedidos)

```python
conn = psycopg2.connect(
            dbname="ecommerce",
            user="postgres",
            password="********",
            host="localhost",
            port="5432"
        )

cursor = conn.cursor()

# Executar a consulta
query = """
SELECT C.Nome, COUNT(P.ID_Pedido) AS Quantidade_Pedidos
FROM Clientes C
JOIN Pedidos P ON C.ID_Cliente = P.ID_Cliente
GROUP BY C.Nome
ORDER BY Quantidade_Pedidos DESC
LIMIT 10;
"""
cursor.execute(query)

# Recuperar e imprimir os resultados
resultados = cursor.fetchall()
for row in resultados:
    print(row)

# Fechar a conexão
cursor.close()
conn.close()
```

13 - Liste os clientes que possuem o mesmo sobrenome.

```python
conn = psycopg2.connect(
            dbname="ecommerce",
            user="postgres",
            password="********",
            host="localhost",
            port="5432"
        )

cursor = conn.cursor()

# Executar a consulta
query = """
SELECT Nome
FROM Clientes C1
WHERE EXISTS (
    SELECT 1
    FROM Clientes C2
    WHERE C1.ID_Cliente <> C2.ID_Cliente
    AND SPLIT_PART(C1.Nome, ' ', -1) = SPLIT_PART(C2.Nome, ' ', -1)
);
"""
cursor.execute(query)

# Recuperar e imprimir os resultados
resultados = cursor.fetchall()
for row in resultados:
    print(row)

# Fechar a conexão
cursor.close()
conn.close()
```

15 - Liste os produtos, sua categoria e uma indicação se estão em estoque ou esgotados (usando CASE).

```python
conn = psycopg2.connect(
            dbname="ecommerce",
            user="postgres",
            password="********",
            host="localhost",
            port="5432"
        )

cursor = conn.cursor()

# Executar a consulta
query = """
SELECT P.Nome, C.Nome AS Categoria,
       CASE
           WHEN P.Quantidade_Estoque > 0 THEN 'Em Estoque'
           ELSE 'Esgotado'
       END AS Status_Estoque
FROM Produtos P
JOIN Categorias C ON P.ID_Categoria = C.ID_Categoria;
"""
cursor.execute(query)

# Recuperar e imprimir os resultados
resultados = cursor.fetchall()
for row in resultados:
    print(row)

# Fechar a conexão
cursor.close()
conn.close()
```
